# Constants of the RLC circuit
Calculate some of the constants related to the RLC circuit.
$$R = 1.5 \Omega$$
$$L = 230 \mu H$$
$$C = 0.5 \mu F$$

```python
R = 1.5
L = 230e-6
C = 0.47e-6
print ("omega=", 1/(L*C)**0.5)
print("f=", 1/(L*C)**0.5/2/np.pi)
print("Q=",  1/(L*C)**0.5*L/R)
print("alpha=", R/2/L)
```
